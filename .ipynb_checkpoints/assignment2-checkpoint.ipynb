1. Prints the answer.1. Area of a Rectangle Formula: A=l*b Inputs: length, breath

```python
l=int(input('enter a l:'))
b=int(input('enter a b:'))
area_of_rectangle=l*b
print('the area_of_rectangle l,b is=area_of_rectangle')
print('the area_of_rectangle {},{} is={}')
print('the area_of_rectangle {}*{} is={}'.format(l,b,area_of_rectangle))
print(f'the area_of_rectangle {l}*{b} is={area_of_rectangle}')
```

2. Perimeter of a Rectangle Formula: P=2(l+w) Inputs: length, width

```python
l=int(input('enter a l:'))
w=int(input('enter a w:'))
perimeter_of_a_Rectangle= 2*(l+w)
perimeter_of_a_Rectangle
print('the perimeter_of_a_Rectangle 2*({}+{}) is={}'.format(l,w,perimeter_of_a_Rectangle))
print(f'the perimeter_of_a_Rectangle 2*({l}+{w}) is={perimeter_of_a_Rectangle}')
```

3. Area of a Square Formula: A=s^2 Input: side

```python
s=int(input('enter a s:'))
area_of_square=s*s
area_of_square
print('the area_of_square {}*{} is={}'.format(s,s,area_of_square))
print(f'the area_of_square {s}*{s} is={area_of_square}')
```

4. Perimeter of a Square Formula: Perimeter = 4 × side Input: side

```python
s=int(input('enter a s:'))
Perimeter_of_a_Square=4 * s
Perimeter_of_a_Square
print('the Perimeter_of_a_Square 4*{} is={}'.format(s,Perimeter_of_a_Square))
print(f'the Perimeter_of_a_Square 4*{s} is={Perimeter_of_a_Square}')
```

5. Area of a Circle Formula: =3.14rr

```python
r=int(input('enter a r:'))
area_of_circle=3.14*r*r
area_of_circle
print('the area_of_circle 3.14*{}*{} is={}'.format(r,r,area_of_circle))
print(f'the area_of_circle 3.14*{r}*{r} is={area_of_circle}')
```

6. Circumference of a Circle Formula:2 * 3.14 * r

```python
r=int(input('enter a r:'))
Circumference_of_a_Circle=2 * 3.14 * r
Circumference_of_a_Circle
print('the Circumference_of_a_Circle {}*3.14*{} is={}'.format(2,r,Circumference_of_a_Circle))
print(f'the Circumference_of_a_Circle {2}*3.14*{r} is={Circumference_of_a_Circle}')
```

7. Area of a Triangle Formula:A = (1/2) × b × h

```python
b=int(input('enter a b:'))
h=int(input('enter a h:'))
Area_of_a_Triangle=(1/2) * b * h
Area_of_a_Triangle
print('the Area_of_a_Triangle 0.5*{}*{} is={}'.format(b,h,Area_of_a_Triangle))
print(f'the Area_of_a_Triangle 0.5*{b}*{h} is={Area_of_a_Triangle}')
```

8. Simple Interest Formula: Simple Interest = (P × R × T) / 100 Inputs: Principal, Rate, Time

```python
p=int(input('enter a p:'))
r=int(input('enter a r:'))
t=int(input('enter a t:'))
Simple_Interest=(p * r * t) / 100
Simple_Interest
print('the Simple_Interest ({}*{}*{})/100 is={}'.format(p,r,t,Simple_Interest))
print(f'the Simple_Interest ({p}*{r}*{t})/100 is={Simple_Interest}')
```

9. Total Amount (Simple Interest) Formula: Amount = Principal + Interest Inputs: Principal, Rate, Time

```python
p=int(input('enter a p:'))
r=int(input('enter a r:'))
t=int(input('enter a t:'))
Simple_Interest=(p * r * t) / 100
Total_Amount=p+Simple_Interest
Total_Amount
print('the Total_Amount {}+{} is={}'.format(p,Simple_Interest,Total_Amount))
print(f'the Total_Amount {p}+{Simple_Interest} is={Total_Amount}')
```

10. Average of 3 Numbers Formula: Average = (a + b + c) / 3 Inputs: three numbers

```python
a=int(input('enter a a:'))
b=int(input('enter a b:'))
c=int(input('enter a c:'))
avg=(a+b+c)/3
avg
print('the avg ({}+{}+{})/3 is={}'.format(a,b,c,avg))
print(f'the avg ({a}+{b}+{c})/3 is={avg}')
```
